```python
%sql
DESCRIBE TABLE workspace.default.gold_echochain_analytics;
```

```python
%sql
SELECT COUNT(*) AS total_rows
FROM workspace.default.gold_echochain_analytics;
```

```python
%sql
SELECT 
    listing_id,
    COUNT(*) AS record_count
FROM workspace.default.gold_echochain_analytics
GROUP BY listing_id
HAVING COUNT(*) > 1;
```

```python
%sql
SELECT
    COUNT(*) AS total_rows,
    COUNT_IF(sku IS NULL) AS null_sku,
    COUNT_IF(listing_id IS NULL) AS null_listing_id,
    COUNT_IF(secondary_market_depreciation IS NULL) AS null_depreciation,
    COUNT_IF(total_component_cost IS NULL) AS null_component_cost,
    COUNT_IF(component_reliability_score IS NULL) AS null_reliability,
    COUNT_IF(resale_retention_score IS NULL) AS null_retention,
    COUNT_IF(circularity_score IS NULL) AS null_circularity
FROM workspace.default.gold_echochain_analytics;
```

```python
%sql
SELECT
    MIN(weighted_score) AS min_weighted_score,
    MAX(weighted_score) AS max_weighted_score,
    MIN(component_reliability_score) AS min_reliability_score,
    MAX(component_reliability_score) AS max_reliability_score,
    MIN(resale_retention_score) AS min_retention_score,
    MAX(resale_retention_score) AS max_retention_score,
    MIN(circularity_score) AS min_circularity_score,
    MAX(circularity_score) AS max_circularity_score,
    MIN(secondary_market_depreciation) AS min_depreciation,
    MAX(secondary_market_depreciation) AS max_depreciation,
    AVG(secondary_market_depreciation) AS avg_depreciation
FROM workspace.default.gold_echochain_analytics;
```
