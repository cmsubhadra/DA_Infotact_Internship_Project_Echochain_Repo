PRODUCTS SILVER LAYER CREATION


```python
from pyspark.sql.functions import col, trim, upper, lower, regexp_replace

products_bronze = spark.table("bronze_products")

display(products_bronze)
```

```python
products_silver = products_bronze.withColumn(
    "sku",
    upper(trim(col("sku")))
)
```

```python
products_silver = products_silver.withColumn(
    "product_name",
    trim(col("product_name"))
)
```

```python
products_silver = products_silver.withColumn(
    "category",
    trim(col("category"))
)
```

```python
display(products_silver)
```

```python
products_silver.printSchema()
```

```python
from pyspark.sql.functions import sum, when

products_silver.select(
    [
        sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
        for c in products_silver.columns
    ]
).show()
```

```python
products_silver.groupBy("sku") \
    .count() \
    .filter(col("count") > 1) \
    .show()
```

```python
products_silver.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("silver_products")
```

```python
display(spark.table("silver_products"))
```

```python
print(
    "Silver products records:",
    spark.table("silver_products").count()
)
```

```python
spark.table("silver_products").printSchema()
```

BOM SILVER LAYER CREATION


```python
from pyspark.sql.functions import col, trim, upper

bom_bronze = spark.table("bronze_bom")

display(bom_bronze)
```

```python
bom_silver = bom_bronze.withColumn(
    "sku",
    upper(trim(col("sku")))
)
```

```python
bom_silver = bom_silver.withColumn(
    "component_id",
    upper(trim(col("component_id")))
)
```

```python
bom_silver = bom_silver.withColumn(
    "component_name",
    trim(col("component_name"))
)
```

```python
display(bom_silver)
```

```python
bom_silver.printSchema()
```

```python
from pyspark.sql.functions import sum, when

bom_silver.select(
    [
        sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
        for c in bom_silver.columns
    ]
).show()
```

```python
bom_silver.groupBy("sku", "component_id") \
    .count() \
    .filter(col("count") > 1) \
    .show()
```

```python
bom_silver.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("silver_bom")
```

```python
display(spark.table("silver_bom"))
```

```python
print(
    "Bronze BOM records:",
    spark.table("bronze_bom").count()
)

print(
    "Silver BOM records:",
    spark.table("silver_bom").count()
)
```
