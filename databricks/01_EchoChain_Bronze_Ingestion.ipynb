PRODUCTS BRONZE LAYER CREATION

```python
products_path ="/Volumes/workspace/default/datasets/products.csv"
products_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(products_path)

display(products_df)
```

```python
print("Number of product records:", products_df.count())
```

```python
print("Product columns:")
print(products_df.columns)
```

```python
products_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("bronze_products")
```

```python
display(spark.table("bronze_products"))
```

```python
print("Bronze product records:", spark.table("bronze_products").count())
```

```python
spark.table("bronze_products").printSchema()
```
